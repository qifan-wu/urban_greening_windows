```python
import geopandas as gpd
import constants
msa_gdf_crsaea = gpd.read_file(constants.MSA_FILE)

regions = constants.CLIMATE_ZONE
import re
def get_region(msa_name, regions=regions):
    pattern = r",\s([^,]+)\sMetro Area$"
    state = re.search(pattern, msa_name).group(1)
    for region, states in regions.items():
        if state in states:
            return region

# save msa with climate zone to shp
msa_gdf_region = msa_gdf_crsaea.copy()
msa_gdf_region['region'] = msa_gdf_region['NAMELSAD'].apply(get_region)
msa_gdf_region.to_file(r'c:\Users\qifanw\Documents\gis\msa\region\msa_all_region.shp', driver="ESRI Shapefile")
msa_gdf_region[['NAMELSAD', 'region']].to_csv(r'c:\Users\qifanw\Documents\data\intermedia\msa_name_region.csv', index=False)
```
