```python
import constants
import rasterio
import xarray as xr
import geopandas as gpd
import numpy as np
import matplotlib.pyplot as plt

msa_ds=gpd.read_file(constants.MSA_REGION_FILE)
```

### MSA area information explore

```python
# Calculate area in square meters
msa_ds['area_km2'] = msa_ds.geometry.area / 1e6
```

```python
msa_ds.sort_values(by='area_km2', ascending=False)
```

```python
import matplotlib.pyplot as plt

# Plot histogram of area in square kilometers
plt.figure(figsize=(10, 6))
msa_ds['area_km2'].hist(bins=100, edgecolor='black')

plt.title('Distribution of Area (km²)')
plt.xlabel('Area (km²)')
plt.ylabel('Number of MSAs')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
msa_ds[['NAMELSAD', 'area_km2']].to_csv(constants.MSA_NAME_AREA_TABLE)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the figure
plt.figure(figsize=(12, 12))

# Use seaborn to create boxplots
sns.boxplot(data=msa_ds, x='region', y='area_km2')

# Customize the plot
plt.xticks(rotation=45)
plt.title('Area Distribution by Region')
plt.xlabel('Region')
plt.ylabel('Area (km²)')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from constants import NEE_MSA_MONTHLY_MEAN_XBASE, MSA_NAME_REGION_TABLE
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(NEE_MSA_MONTHLY_MEAN_XBASE)
row_means = df.iloc[:, 1:].mean(axis=1)
mean_nee = pd.concat([df['NAMELSAD'], row_means.rename('mean_nee')], axis=1)
```

```python
# Read the MSA name-region table
msa_name_region = pd.read_csv(MSA_NAME_REGION_TABLE)

# Merge mean_nee with region info
mean_nee_region = mean_nee.merge(msa_name_region[['NAMELSAD', 'region']], on='NAMELSAD')

# Plot boxplot for mean_nee by region
plt.figure(figsize=(12, 8))
sns.boxplot(data=mean_nee_region, x='region', y='mean_nee')
plt.xticks(rotation=45)
plt.title('Mean NEE by Region')
plt.xlabel('Region')
plt.ylabel('Mean NEE in unit of gC m2-1 day-1')
plt.tight_layout()
plt.show()
```
