```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load data
df = pd.read_csv('../data/intermedia/ml_raw_data.csv') 

# Variables to plot
variables = ['sif', 'tmean', 'ppt', 'tdmean']

# Set up boxplots
fig, axs = plt.subplots(2, 2, figsize=(14, 10))
axs = axs.flatten()

for i, var in enumerate(variables):
    sns.boxplot(data=df, x='region', y=var, ax=axs[i])
    axs[i].set_title(f'{var.upper()} by Region')
    axs[i].set_xlabel('Region')
    axs[i].set_ylabel(var.upper())
    axs[i].tick_params(axis='x', rotation=45)
    axs[i].grid(True)

plt.tight_layout()
plt.show()
```
